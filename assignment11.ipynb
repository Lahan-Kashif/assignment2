```python
import csv
count = 0
with open("cars.csv", "r") as file:
    reader = csv.reader(file)
    next(reader)
    for row in reader:
        count += 1
print("Total records:", count)
```

```python
avg_price_per_year = (
    df.groupby("Year")["MSRP"]
      .mean()
      .round()
      .sort_index()
)
print(avg_price_per_year)
```

```python
import pandas as pd
df = pd.read_csv("cars.csv")
filtered = df[(df["MSRP"] >= 40000) & (df["Popularity"] >= 1000)]
result = (
    filtered.groupby("Make")
    .size()
    .sort_values(ascending=False)
)

print(result)
```
