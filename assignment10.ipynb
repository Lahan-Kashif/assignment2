```python
import csv

file = open("retail supermarket (1).csv", "r")
reader = csv.reader(file)

next(reader)

for row in reader:
    if len(row) < 19:
        print("Skipping row:", row)
        continue

    sales = float(row[17])
    quantity = int(row[18])

    if sales > 500 and quantity > 5:
        print(row)

file.close()
```

```python
import csv

count = 0

with open("retail supermarket (1).csv", "r") as file:
    reader = csv.reader(file)

    header = next(reader)

    # Print column indexes once to find Segment and Region
    for i, col in enumerate(header):
        print(i, col)

    # Replace the indexes below with the correct ones
    segment_index = 6
    region_index = 12

    for row in reader:
        if row[segment_index] == "Corporate" and row[region_index] == "Central":
            count += 1

print("Total records:", count)
```
