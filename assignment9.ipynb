```python
import csv

# Open original file
open_file = open("student_data.csv", "r")
reader = csv.reader(open_file)

# Create new file
new_file = open("students.csv", "w", newline="")
writer = csv.writer(new_file)

# Read header
header = next(reader)
header.append("Percentage")
writer.writerow(header)

# Read each student
for per in reader:
    chem = int(per[2])
    phy = int(per[3])
    math = int(per[4])

    percentage = round((chem + phy + math) / 300 * 100, 2)

    per.append(percentage)

    writer.writerow(per)

# Close files
open_file.close()
new_file.close()

print("Data written successfully!")
```

```python
import csv
open_file = open('student_data.csv',mode = 'r')
reader = csv.reader(open_file)
new_file = open('students.csv',mode= 'w',newline = "")
writer = csv.writer(new_file)
header = next(reader)
header.append("percentage")
header.append("grade")
writer.writerow(header)
#Read each students
for per in reader:
    chem = int(per[2])
    phy = int(per[3])
    math = int(per[4])
    percentage = round((chem + phy + math) / 300 * 100, 2)
    if (percentage >= 80 and percentage <= 100):
         grade= "A"
    elif (percentage >= 70 and percentage <= 79):
        grade= "B"
    elif (percentage >= 60 and percentage <= 69):
        grade= "C"
    elif (percentage >=  50 and percentage <= 59):
     grade= "D"
    else:
     grade= "F"
    per.append(percentage)
    per.append(grade)
    writer.writerow(per)
    # writer.writerow(grade)

open_file.close()
new_file.close()
    
```
