```python
import pandas as pd
import matplotlib.pyplot as plt
df = pd.read_csv("titanic_dataset.csv")
survival_rate = df.groupby("Embarked")["Survived"].mean()
plt.figure(figsize=(8, 5))
survival_rate.plot(kind="bar")
plt.title("Survival Rate by Embarked Port")
plt.xlabel("Embarked")
plt.ylabel("Average of Survived")
plt.xticks(rotation=0)
plt.ylim(0, 1)
plt.show()
highest_port = survival_rate.idxmax()
highest_rate = survival_rate.max()

print(f"Port with the highest survival rate: {highest_port}")
print(f"Survival rate: {highest_rate:.2%}")
```

```python
import pandas as pd
import seaborn as sns
import matplotlib.pyplot as plt
df = pd.read_csv("Titanic_dataset.csv")
df["FamilySize"] = df["SibSp"] + df["Parch"] + 1
plt.figure(figsize=(8,5))
sns.countplot(x="FamilySize", data=df)
plt.title("Number of Passengers by Family Size")
plt.xlabel("Family Size")
plt.ylabel("Number of Passengers")
plt.show()
most_common = df["FamilySize"].mode()[0]
count = df["FamilySize"].value_counts().max()
print("Most common family size:", most_common)
print("Number of passengers:", count)
```

```python
import pandas as pd
import seaborn as sns
import matplotlib.pyplot as plt
df = pd.read_csv("Titanic_dataset.csv")
plt.figure(figsize=(8,6))
sns.boxplot(x="Pclass", y="Fare", hue="Sex", data=df)
plt.title("Fare Distribution by Passenger Class and Sex")
plt.xlabel("Passenger Class (Pclass)")
plt.ylabel("Fare")
plt.legend(title="Sex")
plt.show()
```

```python
import pandas as pd
import matplotlib.pyplot as plt
df = pd.read_csv("titanic_dataset.csv")
table = df.groupby(["Pclass", "Sex", "Survived"]).size().unstack(fill_value=0)
table.plot(kind="bar", stacked=True, figsize=(8,6))
plt.title("Passenger Class & Gender Survival Comparison")
plt.xlabel("Passenger Class and Gender")
plt.ylabel("Number of Passengers")
plt.legend(["Not Survived", "Survived"])

plt.show()
```
