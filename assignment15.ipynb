```python
import sqlite3

conn = sqlite3.connect("factbook.db")
cursor = conn.cursor()

cursor.execute("""
SELECT cities.name, cities.population
FROM cities
JOIN facts
ON cities.facts_id = facts.id
WHERE facts.area > 5000;
""")

rows = cursor.fetchall()

for row in rows:
    print(row)

conn.close()
```

```python
import sqlite3

conn = sqlite3.connect("factbook.db")
cursor = conn.cursor()

cursor.execute("""
SELECT name
FROM facts
WHERE death_rate > 2
  AND birth_rate < 20;
""")

rows = cursor.fetchall()

for row in rows:
    print(row)

conn.close()
```

```python
cursor.execute("""
SELECT name, population_growth
FROM facts
WHERE population_growth BETWEEN 1 AND 5
ORDER BY population_growth DESC;
""")

for row in cursor.fetchall():
    print(row)
    
```

```python
import sqlite3

conn = sqlite3.connect("factbook.db")
cursor = conn.cursor()

cursor.execute("""
SELECT c.name AS capital_city,
       f.death_rate
FROM cities AS c
LEFT JOIN facts AS f
ON c.facts_id = f.id
WHERE c.capital = 1;
""")

rows = cursor.fetchall()

for row in rows:
    print(row)

conn.close()
```
